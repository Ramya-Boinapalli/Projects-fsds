```python
import textwrap


def wrap_text_preserve_newlines(text, width=110):
  #split
    lines = text.split('\n')
    wrapped_lines = [textwrap.fill(line, width=width) for line in lines]
    wrapped_text = '\n'.join(wrapped_lines)

    return wrapped_lines

```

```python
print(wrap_text_preserve_newlines(str(documents[0])))
```

```python
from langchain.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain.text_splitter import CharacterTextSplitter

# Install necessary packages for embeddings if not already installed
!pip install -U langchain-huggingface sentence-transformers

# Re-split documents to ensure 'docs' is a list of Document objects
# Assuming 'documents' and 'text_splitter' are defined from previous cells
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=0)
docs = text_splitter.split_documents(documents)

# Initialize the embedding model
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Create the FAISS vector store
db = FAISS.from_documents(docs, embeddings)
```

```python
query = " expalin about cricket give in paragraph?"
search_results = db.similarity_search(query)
```
